```python
import yfinance as yf
import pandas as pd
import mplfinance as mpf
import talib
import time
import os
```

```python
#Generating Patterns

tickers = ['AAPL', 'GOOGL', 'MSFT', 'TSLA', 'NVDA', 'META', 'JPM', 'NFLX', 'AMD']

# Create directory to save hammer images
output_dir = "Inverted_Hammer"
os.makedirs(output_dir, exist_ok=True)

image_count = 1  # To track file numbering

for ticker in tickers:
    raw = yf.download(ticker, period='730d', interval='60m')[['Open', 'High', 'Low', 'Close']]
    
    if isinstance(raw.columns, pd.MultiIndex):
        raw.columns = raw.columns.droplevel(1)
    
    df = raw[['Open', 'High', 'Low', 'Close']].copy()
    print(f"Downloaded data for {ticker}")

    df["Inverted_Hammer"] = talib.CDLINVERTEDHAMMER(
        df['Open'].values.ravel(),
        df['High'].values.ravel(),
        df['Low'].values.ravel(),
        df['Close'].values.ravel()
    )

    inv_signals = df[df["Inverted_Hammer"] != 0].index
    
    for timestamp in inv_signals:
        i = df.index.get_loc(timestamp)

        if i < 10 or i + 10 >= len(df):
            continue

        segment = df.iloc[i - 10:i + 10][['Open', 'High', 'Low', 'Close']] \
            .dropna() \
            .applymap(float)

        # Save the candlestick plot as an image
        save_path = os.path.join(output_dir, f"IHammer_{image_count}.png")
        mpf.plot(
            segment,
            type='candle',
            style='yahoo',
            show_nontrading=False,
            savefig=dict(fname=save_path, dpi=200, bbox_inches='tight', pad_inches=0.02)
        )
        print(f"Saved: {save_path}")
        image_count += 1

```

```python
window = 5          # block +/- N candles around any detected pattern
segment_len = 20    # number of candles per saved image (use 10 for i-5:i+5)
max_images = 3000     # limit total saved images
style = 'yahoo'
import os
import pandas as pd
import yfinance as yf
import talib
import mplfinance as mpf

# Generating NOT patterns
tickers = [

    # --- Tech ---
    'AAPL', 'MSFT', 'NVDA', 'TSLA', 'META', 'AMZN', 'GOOGL', 'GOOG', 'AMD', 'INTC',
    'ORCL', 'CRM', 'ADBE', 'CSCO', 'SHOP', 'UBER', 'PYPL', 'NFLX', 'AVGO',
    'ZM', 'DOCU', 'SNOW', 'PLTR', 'DDOG', 'OKTA', 'CRWD', 'ZS', 'NET', 'MDB',

    # --- Semiconductors ---
    'QCOM', 'TXN', 'MU', 'TSM', 'ASML', 'AMAT', 'LRCX', 'KLAC', 'NXPI', 'ON', 'MRVL',

    # --- Financials ---
    'JPM', 'BAC', 'C', 'GS', 'MS', 'BLK', 'V', 'MA', 'AXP', 'SCHW', 'COF', 'ALLY',

    # --- Industrials & Aerospace ---
    'BA', 'CAT', 'GE', 'HON', 'DE', 'NOC', 'LMT', 'RTX', 'TDG', 'GD', 'HEI', 'TXT',

    # --- Consumer & Retail ---
    'DIS', 'HD', 'LOW', 'NKE', 'SBUX', 'MCD', 'TGT', 'COST', 'WMT', 'YUM', 'KO', 'PEP',
    'KHC', 'PG', 'CL', 'EL', 'KMB',
    

    # --- Energy ---
    'XOM', 'CVX', 'SLB', 'COP', 'OXY', 'PSX', 'EOG', 'FANG', 'MPC', 'VLO',

    # --- Pharma & Healthcare ---
    'PFE', 'MRNA', 'JNJ', 'UNH', 'LLY', 'ABBV', 'BMY', 'MRK', 'GILD', 'REGN', 'BIIB',

    # --- ETFs ---
    'SPY', 'QQQ', 'IWM', 'DIA', 'XLK', 'XLF', 'XLE', 'XLV', 'XLY', 'XLI', 'ARKK',
    'SOXX', 'SMH', 'XLRE', 'XLP', 'XLB', 'XLC',

    # --- Crypto-related equities ---
    'COIN', 'RIOT', 'MARA', 'MSTR', 'HUT', 'BTBT', 'SI',

    # --- Automakers / EV ---
    'F', 'GM', 'RIVN', 'LCID', 'NIO', 'XPEV', 'LI', 'BYDDF',

    # --- International Large Caps ---
    'BABA', 'JD', 'PDD', 'NTES', 'TCEHY', 'BIDU', 'INFY', 'MELI', 'SHOP',
     'BP', 'RIO', 'BHP', 'SHEL'
]

output_dir = "NHammer_images"
os.makedirs(output_dir, exist_ok=True)

image_count = 1  # To track file numbering
max_images = 3000  # Limit total saved images

for ticker in tickers:
    if image_count > max_images:
        break  # Stop outer loop if we've reached the limit

    raw = yf.download(ticker, period='1y', interval='1h')[['Open', 'High', 'Low', 'Close']]
    if isinstance(raw.columns, pd.MultiIndex):
        raw.columns = raw.columns.droplevel(1)

    df = raw[['Open', 'High', 'Low', 'Close']].copy()
    print(f"Downloaded data for {ticker}")

    # -------------------- TA-Lib pattern detection (expanded) --------------------
    # Patterns from your bullish/bearish list that TA-Lib supports.
    # (Side-by-Side Lines and Mat Hold are not available in TA-Lib, so omitted.)
    pattern_funcs = {
    # === Bullish ===
    "CDLHAMMER": talib.CDLHAMMER,                         # Hammer
    "CDLINVERTEDHAMMER": talib.CDLINVERTEDHAMMER,         # Inverted Hammer
    "CDLENGULFING": talib.CDLENGULFING,                   # Engulfing (+100 bullish, -100 bearish)
    "CDLPIERCING": talib.CDLPIERCING,                     # Piercing Line
    "CDLMORNINGSTAR": talib.CDLMORNINGSTAR,               # Morning Star
    "CDLMORNINGDOJISTAR": talib.CDLMORNINGDOJISTAR,       # Morning Doji Star
    "CDL3WHITESOLDIERS": talib.CDL3WHITESOLDIERS,         # Three White Soldiers
    "CDLHARAMI": talib.CDLHARAMI,                         # Harami (+ bullish, - bearish)
    "CDLHARAMICROSS": talib.CDLHARAMICROSS,               # Harami Cross (+ bullish, - bearish)
    "CDLMATCHINGLOW": talib.CDLMATCHINGLOW,               # (closest TA‑Lib analog to Tweezer Bottom)
    "CDLTASUKIGAP": talib.CDLTASUKIGAP,                   # Tasuki Gap (+ bullish, - bearish)
    "CDLGAPSIDESIDEWHITE": talib.CDLGAPSIDESIDEWHITE,     # Side‑by‑Side White Lines (+ up‑gap bull, - down‑gap bear)
    "CDLDRAGONFLYDOJI": talib.CDLDRAGONFLYDOJI,           # Dragonfly Doji
    "CDL3INSIDE": talib.CDL3INSIDE,                       # Three Inside (Up/Down by sign)
    "CDLABANDONEDBABY": talib.CDLABANDONEDBABY,           # Abandoned Baby (+ bullish, - bearish)

    # === Bearish ===
    "CDLSHOOTINGSTAR": talib.CDLSHOOTINGSTAR,             # Shooting Star
    "CDLHANGINGMAN": talib.CDLHANGINGMAN,                 # Hanging Man
    # (CDLENGULFING already included; use sign < 0 for bearish engulfing)
    "CDLDARKCLOUDCOVER": talib.CDLDARKCLOUDCOVER,         # Dark Cloud Cover
    "CDLEVENINGSTAR": talib.CDLEVENINGSTAR,               # Evening Star
    "CDLEVENINGDOJISTAR": talib.CDLEVENINGDOJISTAR,       # Evening Doji Star
    "CDLUPSIDEGAP2CROWS": talib.CDLUPSIDEGAP2CROWS,       # Upside Gap Two Crows
    # (CDLHARAMI / CDLHARAMICROSS already included; use sign < 0 for bearish)
    "CDLCOUNTERATTACK": talib.CDLCOUNTERATTACK,           # Counterattack (+ bull, - bear)
    "CDLONNECK": talib.CDLONNECK,                         # On‑Neck Line (bearish continuation)
    # (CDLTASUKIGAP already included; use sign < 0 for bearish)
    "CDLTRISTAR": talib.CDLTRISTAR,                       # Tristar (+ bull, - bear)
    "CDLSTALLEDPATTERN": talib.CDLSTALLEDPATTERN,         # Stalled Pattern (bearish)
}
    

    O = df['Open'].values.ravel()
    H = df['High'].values.ravel()
    L = df['Low'].values.ravel()
    C = df['Close'].values.ravel()

    # Create one column per pattern
    for name, func in pattern_funcs.items():
        df[name] = func(O, H, L, C)

    # Build a combined mask of ANY detected pattern (non-zero)
    any_pattern_mask = df[list(pattern_funcs.keys())].ne(0).any(axis=1)
    # ---------------------------------------------------------------------------

    # Block any rows within ±5 candles of a true pattern (kept your ±5 logic)
    blocked = set()
    for ts in df.index[any_pattern_mask]:
        j = df.index.get_loc(ts)
        for k in range(j - 5, j + 6):
            if 0 <= k < len(df):
                blocked.add(df.index[k])

    # Keep only bars with NO pattern and NOT in blocked zone
    clean_none_indices = [ts for ts in df.index[~any_pattern_mask] if ts not in blocked]

    # Save images
    for timestamp in clean_none_indices:
        if image_count > max_images:
            break  # Stop inner loop if we've reached the limit

        i = df.index.get_loc(timestamp)
        if i < 10 or i + 10 >= len(df):
            continue

        segment = df.iloc[i - 10:i + 10][['Open', 'High', 'Low', 'Close']] \
                     .dropna() \
                     .astype(float)

        save_path = os.path.join(output_dir, f"none_{image_count}.png")
        mpf.plot(
            segment,
            type='candle',
            style='yahoo',
            show_nontrading=False,
            savefig=save_path
        )
        print(f"Saved: {save_path}")
        image_count += 1

print(f"Done. Saved {image_count-1} NONE images.")
```
